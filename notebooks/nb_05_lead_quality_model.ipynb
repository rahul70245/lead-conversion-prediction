```python
# for working part
import numpy as np
import pandas as pd


# for ml
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score,confusion_matrix,precision_score,recall_score,f1_score,roc_auc_score
```

```python
# reading the parquet file created in notebook 2

lead_conversion_data = pd.read_parquet("C:\\Users\\hp5cd\\OneDrive\\Desktop\\Python\\lead-conversion-prediction\\data\\processed_data.parquet")
```

```python
# extracting important columns for this model
# as this will be used when creating leads we will have columns only that can be there after creating leads (x1)
# and when assigned we will have the columns that are in x2 


x1 = lead_conversion_data[['lead_source']]
x2 = lead_conversion_data[['lead_source','owner','assigned_month','assigned_year']]

y = lead_conversion_data['converted']
```

```python
## stratify is used to keep the distribution of output variables same in test and train split
## test size 0.2 means 80% training split of data and 20% will be in test split


x1_train, x1_test, y1_train, y1_test = train_test_split(x1,y,random_state=42,test_size=0.2,stratify=y)

x2_train, x2_test, y2_train, y2_test = train_test_split(x2,y,random_state=42,test_size=0.2,stratify=y)
```

```python
# categorical columns

categorical_x1 = ['lead_source']
categorical_x2 = ['lead_source','owner','assigned_month','assigned_year']
```

```python
# preprocessing

preprocessor1 = ColumnTransformer(transformers=[('cat',OneHotEncoder(handle_unknown = 'ignore'),categorical_x1)])

# models
models = {'LR':LogisticRegression(max_iter=1000),'RF':RandomForestClassifier(random_state=42),'XGB': XGBClassifier()}


# pipeline with multiple model

for name,model in models.items():
    pipeline = Pipeline([
                        ('preprocessor', preprocessor1),
                         ('model', model)
                         ])
    
    pipeline.fit(x1_train,y1_train)

    y1_pred = pipeline.predict(x1_test)
    y1_pred_prob = pipeline.predict_proba(x1_test)[:,1]
    
    print("model", name)

    print(" ")
    print("accuracy_score: ", accuracy_score(y1_test,y1_pred))
    print("confusion_matrix: ", confusion_matrix(y1_test,y1_pred))
    print("precision_score: ", precision_score(y1_test,y1_pred))
    print("recall_score: ", recall_score(y1_test,y1_pred))
    print("f1_score: ", f1_score(y1_test,y1_pred))
    print("roc_auc_score: ", roc_auc_score(y1_test,y1_pred_prob))

    print(" ")





```

```python
# SECOND MODEL

# preprocessing
preprocessor2 = ColumnTransformer(transformers=[('cat',OneHotEncoder(handle_unknown='ignore'),categorical_x2)])


# definind models to train
models2 = {'LR':LogisticRegression(max_iter=1000),
            'RF':RandomForestClassifier(n_estimators=20,max_depth=10,n_jobs=-1,random_state=42), 
            'XGB': XGBClassifier(n_estimators=50,max_depth=5,learning_rate=0.1,n_jobs=-1,random_state=42)}

for name2,model2 in models2.items():
    pipeline2 = Pipeline([('preprocessor',preprocessor2),
                          ('model',model2)])
    
    pipeline2.fit(x2_train,y2_train)

    y2_pred = pipeline2.predict(x2_test)
    y2_pred_prob = pipeline2.predict_proba(x2_test)[:,1]
    
    print("model", name2)

    print(" ")
    print("accuracy_score: ", accuracy_score(y2_test,y2_pred))
    print("confusion_matrix: ", confusion_matrix(y2_test,y2_pred))
    print("precision_score: ", precision_score(y2_test,y2_pred))
    print("recall_score: ", recall_score(y2_test,y2_pred))
    print("f1_score: ", f1_score(y2_test,y2_pred))
    print("roc_auc_score: ", roc_auc_score(y2_test,y2_pred_prob))

    print(" ")


```
